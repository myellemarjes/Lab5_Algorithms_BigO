```python
import time

# Student-specific inputs
LAST_NAME = input("Marjes").strip().upper()
STUDENT_ID = input("TUPM-26-1948").strip()
FAVORITE_ARTIST = input("The Ridleys").strip().upper()

# Generate student-specific values
SEED_DIGIT = next(
    (int(char) for char in reversed(STUDENT_ID) if char.isdigit()),
    0
)
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))

ARTIST_SUM = sum(
    ord(char) - ord("A") + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
)

print("\n--- Student-Specific Values ---")
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)

# Generate unique incident records
n = 20
start = max(ID_SUM, 1) * 100 + SEED_DIGIT
original = list(range(start, start + n))

# Rotate the list using SEED_DIGIT
rotation = SEED_DIGIT % n
rotated = original[rotation:] + original[:rotation]

# Generate a target incident using ID_SUM and ARTIST_SUM
target_index = (ID_SUM + ARTIST_SUM) % n
target = original[target_index]

print("\nOriginal Dataset:", original)
print("Rotation Point:", rotation)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)

# Linear search
comparisons = 0
found_index = -1

start_time = time.perf_counter()

for i, value in enumerate(rotated):
    comparisons += 1
    if value == target:
        found_index = i
        break

end_time = time.perf_counter()

print("\n--- Search Results ---")
if found_index != -1:
    print("Target found at rotated index:", found_index)
else:
    print("Target incident not found.")

print("Comparisons:", comparisons)
print(f"Execution Time: {(end_time - start_time) * 1_000_000:.2f} microseconds")
print("Time Complexity: O(n)")
print("Search Auxiliary Space: O(1)")
```

```python
import random
import time

# Student-specific inputs
STUDENT_ID = "TUPM-26-1948"
FAVORITE_ARTIST = "The Ridleys".upper()

# Generate the same set of order numbers
n = 100
orders = list(range(1, n + 1))

# Create the three input conditions
sorted_orders = orders.copy()
reverse_orders = orders[::-1]

random.seed(SEED_DIGIT + ARTIST_SUM)
random_orders = orders.copy()
random.shuffle(random_orders)

# Bubble Sort
def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        swapped = False

        for j in range(len(arr) - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements


# Selection Sort
def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 3

    return arr, comparisons, movements


# Insertion Sort
def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements


algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

conditions = {
    "Sorted": sorted_orders,
    "Reverse": reverse_orders,
    "Random": random_orders
}

print("\n========== E-COMMERCE SORTING RESULTS ==========")

for condition, dataset in conditions.items():
    print(f"\n--- {condition} Input ---")
    print("Original Dataset (first 15):", dataset[:15])

    for name, algorithm in algorithms.items():
        start_time = time.perf_counter()

        result, comparisons, movements = algorithm(dataset)

        end_time = time.perf_counter()
        elapsed = end_time - start_time

        print(f"\n{name}")
        print("Sorted Correctly:", result == sorted(dataset))
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print(f"Execution Time: {elapsed * 1_000_000:.2f} microseconds")
        print("Time Complexity: O(n^2)")
        print("Auxiliary Space: O(1)")
        print("Sorted Output (first 15):", result[:15])
```
